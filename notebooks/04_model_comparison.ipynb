# Model Comparison and Forecast Visualisation

This notebook compares multiple forecasting models on UK electricity demand and visualises their predictions against actual demand.


```python
import pandas as pd
import matplotlib.pyplot as plt

from pathlib import Path

from src.evaluation.model_registry import list_available_models, load_model_forecast


```

```python
plt.style.use("seaborn-v0_8")

data_path = Path("../data/processed/modelling_dataset.csv")
results_path = Path("../reports/model_results_all.csv")

df = pd.read_csv(data_path, parse_dates=["datetime"]).set_index("datetime")
df.head()
```

# 2 load model comparison table

```python
results = pd.read_csv(results_path)
results.sort_values("mae")
```

# 3 — MAE bar chart

```python
plt.figure(figsize=(8, 4))
results.sort_values("mae").set_index("model")["mae"].plot(kind="bar")
plt.ylabel("MAE")
plt.title("Model comparison (lower is better)")
plt.show()
```

# 4 — list available forecast files

```python
models = list_available_models("../reports/forecasts")
models
```

# 5 — plot actual vs one model (e.g. ARIMA)

```python
model_name = "ARIMA"  # change to any name from `models`

forecast_df = load_model_forecast(model_name, "../reports/forecasts")

horizon = 500  # last N points

plt.figure(figsize=(14, 5))
df["demand"].iloc[-horizon:].plot(label="Actual", color="black")
forecast_df["forecast"].iloc[-horizon:].plot(label=model_name)
plt.legend()
plt.title(f"Actual vs {model_name} forecast (last {horizon} points)")
plt.show()
```

# 6 — compare multiple models on the same plot

```python
compare_models = ["ARIMA", "SARIMA", "SARIMAX"]  # adjust based on what you have

horizon = 500

plt.figure(figsize=(14, 5))
df["demand"].iloc[-horizon:].plot(label="Actual", color="black")

for m in compare_models:
    f = load_model_forecast(m, "../reports/forecasts")
    f["forecast"].iloc[-horizon:].plot(label=m)

plt.legend()
plt.title(f"Actual vs multiple model forecasts (last {horizon} points)")
plt.show()
```
